```python
import pandas as pd
```

```python
# 1. Carica il file MIB e correggi la colonna delle Date
df_mib = pd.read_excel('./Dataset/Stocks/MIB_Stocks.xlsx')
df_mib = df_mib.rename(columns={'Name': 'Date'})
df_mib['Date'] = pd.to_datetime(df_mib['Date'])
```

```python
# 2. Estrai i dati per una specifica azienda e standardizza le colonne
df_a2a = df_mib[['Date', 'A2A', 'A2A - PRICE HIGH', 'A2A - PRICE LOW', 'A2A - OPENING PRICE', 'A2A - TURNOVER BY VOLUME']].copy()
df_a2a.columns = ['Date', 'Close', 'High', 'Low', 'Open', 'Volume']
df_a2a['Company'] = 'A2A'
```

```python
# 3. Carica i dati Macro (es. GVZ) formattando la data
df_gvz = pd.read_csv('GVZ.csv')
df_gvz = df_gvz.rename(columns={'DATE': 'Date'})
df_gvz['Date'] = pd.to_datetime(df_gvz['Date'])

# 4. Unisci i dati dell'azienda con l'indicatore macro e gestisci le festività sfalsate (ffill)
df_final = pd.merge(df_a2a, df_gvz, on='Date', how='left')
df_final = df_final.fillna(method='ffill')

# Risultato: df_final è pronto per estrarre le finestre da X giorni 
# precedenti ad ogni singola trimestrale di A2A!
```

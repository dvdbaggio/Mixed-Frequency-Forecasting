```python
def load_and_prep(file_path, date_col, value_cols, new_col_names=None, delimiter=','):
    if file_path.endswith('.xlsx'):
        df = pd.read_excel(file_path)
    else:
        df = pd.read_csv(file_path, sep=delimiter)
    df = df.rename(columns={date_col: 'Date'})
    df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

    cols_to_keep = ['Date'] + value_cols
    df = df[cols_to_keep].dropna(subset=['Date'])
    
    if new_col_names is not None:
        rename_dict = dict(zip(value_cols, new_col_names))
        df = df.rename(columns=rename_dict)
    
    df = df.set_index('Date')
    df = df.sort_index() 
    return df
```

```python
def process_fundamental_file(file_path):
    df = pd.read_excel(file_path)

    df = df.rename(columns={df.columns[0]: 'Report_Serial'})
    df = df.dropna(how='all').dropna(axis=1, how='all')
    
    # Melt turns columns into rows: Report_Serial | Company_Metric | Value
    df_long = df.melt(id_vars=['Report_Serial'], var_name='Company_Metric', value_name='Value')
    

    df_long = df_long.dropna(subset=['Value'])
    df_long = df_long[~df_long['Value'].astype(str).str.contains('ERROR|ER:', na=False, case=False)]
    df_long['Value'] = pd.to_numeric(df_long['Value'], errors='coerce')
    df_long = df_long.dropna(subset=['Value'])
    
    known_metrics = [
        'EBIT & DEPRECIATION',
        'NET INCOME - BASIC',
        'TOTAL ASSETS',
        'TOTAL DEBT',
        'CASH & SHORT TERM INVESTMENTS',
        'NET SALES OR REVENUES',
        'OPERATING EXPENSES - TOTAL',
        'FREE CASH FLOW PER SHARE',
    ]
    
    def extract_info(col_name):
        for metric in known_metrics:
            if str(col_name).endswith(metric):
                company = str(col_name).replace(f" - {metric}", "").strip()
                return pd.Series([company, metric])
        return pd.Series([np.nan, np.nan])
    

    df_long[['Company', 'Metric']] = df_long['Company_Metric'].apply(extract_info)
    df_long = df_long.dropna(subset=['Company', 'Metric'])

    df_clean = df_long.pivot_table(index=['Report_Serial', 'Company'],
                                   columns='Metric', 
                                   values='Value').reset_index()
    
    df_clean = df_clean.rename(columns={
        'EBIT & DEPRECIATION': 'EBITDA',
        'NET INCOME - BASIC': 'Net_Income',
        'TOTAL ASSETS': 'Total_Assets',
        'TOTAL DEBT': 'Total_Debt',
        'CASH & SHORT TERM INVESTMENTS': 'Cash',
        'NET SALES OR REVENUES': 'Net_Sales',
        'OPERATING EXPENSES - TOTAL': 'Operating_Expenses',
        'FREE CASH FLOW PER SHARE': 'FCF_Per_Share',
    })
    
    return df_clean
```

```python
df_mib = process_fundamental_file('./Raw/Targets/MIB_Target.xlsx')
df_mid = process_fundamental_file('./Raw/Targets/MID_Target.xlsx')
df_small = process_fundamental_file('./Raw/Targets/SMALL_Target.xlsx')

df_targets = pd.concat([df_mib, df_mid, df_small], ignore_index=True)
```

```python
def report_serial_to_year_end(v):
    """Convert a report-date value to end-of-that-year 'YYYY-12-31'.

    Accepts an Excel-serial date (int/float; common for MIB/MID files), a
    pandas Timestamp (SMALL files), or a stray string (returns NaN, later
    dropped). Year is taken from the report date verbatim.
    """
    if pd.isna(v):
        return np.nan
    if isinstance(v, (pd.Timestamp,)):
        ts = v
    elif isinstance(v, (int, float, np.integer, np.floating)):
        ts = pd.Timestamp('1899-12-30') + pd.Timedelta(days=int(v))
    else:
        return np.nan
    return f"{ts.year}-12-31"
```

```python
df_targets['Date'] = pd.to_datetime(df_targets['Report_Serial'].apply(report_serial_to_year_end))
df_targets = df_targets.dropna(subset=['Date'])
df_targets = df_targets.drop(columns=['Report_Serial'])

df_targets['ROA'] = df_targets['Net_Income'] / df_targets['Total_Assets']
cols = ['Date', 'Company', 'EBITDA', 'Net_Income', 'ROA',
        'Total_Assets', 'Total_Debt', 'Cash',
        'Net_Sales', 'Operating_Expenses', 'FCF_Per_Share']
df_targets = df_targets[cols]
df_targets = df_targets.sort_values(by=['Company', 'Date']).reset_index(drop=True)
```

```python
df_targets['Year'] = df_targets['Date'].dt.year
df_targets = df_targets.drop_duplicates(subset=['Company', 'Year'], keep='last')
df_targets = df_targets.drop(columns=['Year']).reset_index(drop=True)

df_targets.to_csv('./Processed/Targets_yoy.csv', index=False)
print(f"Extraction complete! Retained {len(df_targets)} annual reports.")
```

# Prior-Year Fundamentals (Static Features)

For each (Company, year) row, attach the prior calendar year's fundamentals
(EBITDA, Net_Income, ROA, Total_Assets, Total_Debt, Cash, Net_Sales,
Operating_Expenses, FCF_Per_Share) via an explicit year-offset merge so a
missing year Y-1 yields NaN rather than silently pulling an older year.
From these prior-year values, compute the 9 model-ready static features the
LSTM consumes: symlog-transformed prior EBITDA / Net_Income / Net_Sales /
Operating_Expenses / FCF_Per_Share, prior ROA, leverage, cash ratio, and
firm size.

Rows missing the prior year's fundamentals are left as NaN here; the model
notebook drops them at load time (preserves existing behavior).

```python
df_targets['Year'] = df_targets['Date'].dt.year

_prior_rename = {
    'EBITDA': 'Prior_EBITDA', 'Net_Income': 'Prior_Net_Income',
    'ROA': 'Prior_ROA_raw', 'Total_Assets': 'Prior_Total_Assets',
    'Total_Debt': 'Prior_Total_Debt', 'Cash': 'Prior_Cash',
    'Net_Sales': 'Prior_Net_Sales', 'Operating_Expenses': 'Prior_Operating_Expenses',
    'FCF_Per_Share': 'Prior_FCF_Per_Share',
}
prior = df_targets[['Company', 'Year'] + list(_prior_rename.keys())].copy()
prior['Year'] = prior['Year'] + 1
prior = prior.rename(columns=_prior_rename)

df_targets = df_targets.merge(prior, on=['Company', 'Year'], how='left')

EPS = 1e-8
df_targets['Prior_EBITDA_SL']     = np.sign(df_targets['Prior_EBITDA']) * np.log1p(np.abs(df_targets['Prior_EBITDA']))
df_targets['Prior_NI_SL']         = np.sign(df_targets['Prior_Net_Income']) * np.log1p(np.abs(df_targets['Prior_Net_Income']))
df_targets['Prior_ROA']           = df_targets['Prior_ROA_raw']
df_targets['Leverage']            = df_targets['Prior_Total_Debt'] / (np.abs(df_targets['Prior_Total_Assets']) + EPS)
df_targets['Cash_Ratio']          = df_targets['Prior_Cash'] / (np.abs(df_targets['Prior_Total_Assets']) + EPS)
df_targets['Size_SL']             = np.sign(df_targets['Prior_Total_Assets']) * np.log1p(np.abs(df_targets['Prior_Total_Assets']))
df_targets['Prior_Net_Sales_SL']     = np.sign(df_targets['Prior_Net_Sales']) * np.log1p(np.abs(df_targets['Prior_Net_Sales']))
df_targets['Prior_OpEx_SL']          = np.sign(df_targets['Prior_Operating_Expenses']) * np.log1p(np.abs(df_targets['Prior_Operating_Expenses']))
df_targets['Prior_FCF_Per_Share_SL'] = np.sign(df_targets['Prior_FCF_Per_Share']) * np.log1p(np.abs(df_targets['Prior_FCF_Per_Share']))

df_targets = df_targets.drop(columns=['Year', 'Prior_EBITDA', 'Prior_Net_Income',
                                       'Prior_ROA_raw', 'Prior_Total_Assets',
                                       'Prior_Total_Debt', 'Prior_Cash',
                                       'Prior_Net_Sales', 'Prior_Operating_Expenses',
                                       'Prior_FCF_Per_Share'])

_new_static = ['Prior_EBITDA_SL', 'Prior_NI_SL', 'Prior_ROA',
               'Leverage', 'Cash_Ratio', 'Size_SL',
               'Prior_Net_Sales_SL', 'Prior_OpEx_SL', 'Prior_FCF_Per_Share_SL']
n_with = df_targets[_new_static].notna().all(axis=1).sum()
n_without = len(df_targets) - n_with
print(f"Targets with all 9 prior-year static features: {n_with}/{len(df_targets)} "
      f"({100.0 * n_with / max(1, len(df_targets)):.2f}%).")
print(f"Targets missing prior-year fundamentals (left as NaN): {n_without}.")

df_targets.to_csv('./Processed/Targets_yoy.csv', index=False)
```

# Merging Datasets

```python
df_stocks = pd.read_csv('./Processed/Stocks.csv')
df_stocks['Date'] = pd.to_datetime(df_stocks['Date'])

df_macro = pd.read_csv('./Processed/Macro.csv')
df_macro['Date'] = pd.to_datetime(df_macro['Date'])

df_targets = pd.read_csv('./Processed/Targets_yoy.csv')
df_targets['Date'] = pd.to_datetime(df_targets['Date'])
```

```python
df_daily = pd.merge(df_stocks, df_macro, on='Date', how='inner')

daily_keys = (
    df_daily[['Company', 'Date']]
    .drop_duplicates()
    .rename(columns={'Date': 'Date_daily'})
    .sort_values(by=['Date_daily', 'Company'])
)

targets_sorted = (
    df_targets
    .rename(columns={'Date': 'Date_target'})
    .sort_values(by=['Date_target', 'Company'])
)

df_targets_aligned = pd.merge_asof(
    targets_sorted,
    daily_keys,
    left_on='Date_target',
    right_on='Date_daily',
    by='Company',
    direction='backward'
)

df_targets_aligned = (
    df_targets_aligned
    .dropna(subset=['Date_daily'])
    .drop(columns=['Date_target'])
    .rename(columns={'Date_daily': 'Date'})
)

df_final = pd.merge(df_daily, df_targets_aligned, on=['Company', 'Date'], how='left')

df_final = df_final.sort_values(by=['Company', 'Date']).reset_index(drop=True)
df_final.to_csv('./Processed/Merged_Dataset_yoy.csv', index=False)
```

# Macro Rozes

```python
df = pd.read_csv("./Processed/Merged_Dataset_yoy.csv")
df["Date"] = pd.to_datetime(df["Date"])

macro = pd.read_csv("./Rozes/macro_daily.csv")
macro["Date"] = pd.to_datetime(macro["date"])

macro_features = [
    "T5YIE",
    "YC__level",
    "YC__slope",
    "YC__curv",
    "PCOPPUSDM",
    "PALUMUSDM",
    "NFCI",
    "ff_Mkt-RF",
    "ff_HML",
    "ff_Mom"
]

macro = macro[["Date"] + macro_features]

df = df.merge(
    macro,
    on="Date",
    how="left"
)

df.to_csv("./Processed/Merged_Dataset_yoy.csv", index=False)
```

# Checking for Missing Values in Targets

```python
df = pd.read_csv("./Processed/Merged_Dataset_yoy.csv")
missing_ebitda = df[df["EBITDA"].isna() & df["Net_Income"].notna()]
missing_net_income = df[df["Net_Income"].isna() & df["EBITDA"].notna()]
missing_roa = df[df["ROA"].isna() & df["Net_Income"].notna()]

print(f"Rows without EBITDA: {len(missing_ebitda)}")
print(f"Rows without Net Income: {len(missing_net_income)}")
print(f"Rows without ROA: {len(missing_roa)}")
for col in ["Net_Sales", "Operating_Expenses", "FCF_Per_Share"]:
    n_missing = df[col].isna().sum()
    print(f"Rows without {col}: {n_missing}")
```

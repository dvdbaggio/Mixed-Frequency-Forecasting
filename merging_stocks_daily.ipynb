```python
import pandas as pd
```

```python
df_mib = pd.read_excel('./Dataset/Stocks/MIB_Stocks.xlsx')
df_mib = df_mib.rename(columns={'Name': 'Date'})
df_mib['Date'] = pd.to_datetime(df_mib['Date'])

stocks = []
stocks_names = [col for col in df_mib.columns if ' - ' not in col and col != 'Date']

for stock in stocks_names:
    col_close = stock
    col_high = f"{stock} - PRICE HIGH"
    col_low = f"{stock} - PRICE LOW"
    col_open = f"{stock} - OPENING PRICE"
    col_volume = f"{stock} - TURNOVER BY VOLUME"
    
    columns = [col_close, col_high, col_low, col_open, col_volume]
    if all(col in df_mib.columns for col in columns):
        df_temp = df_mib[['Date'] + columns].copy()
        df_temp.columns = ['Date', 'Close', 'High', 'Low', 'Open', 'Volume']
        df_temp['Company'] = stock
        stocks.append(df_temp)

df_mib_complete = pd.concat(stocks, ignore_index=True)
df_mib_complete = df_mib_complete.dropna(subset=['Close', 'High', 'Low', 'Open', 'Volume'])

#df_mib_complete.to_csv('./Processed/MIB_Cleaned.csv', index=False)
```
